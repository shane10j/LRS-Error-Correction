# Support Sanity

Thin notebook for auditing support feature construction before model tuning.

```python
from pathlib import Path
import subprocess
import sys

cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd if (cwd / 'scripts').exists() else cwd.parent
PYTHON = Path('/Users/shanejayasundera/anaconda3/envs/lrs_err_correct_env/bin/python')
CONFIG = PROJECT_ROOT / 'configs' / 'debug_tiny.yaml'
OUTPUT = PROJECT_ROOT / 'outputs' / 'support_sanity' / 'support_sanity.json'

print('PROJECT_ROOT =', PROJECT_ROOT)
print('PYTHON =', PYTHON)
print('CONFIG =', CONFIG)
print('OUTPUT =', OUTPUT)
```

```python
env = dict(**__import__('os').environ)
env['PYTHONPATH'] = str(PROJECT_ROOT / 'src')

command = [
    str(PYTHON),
    str(PROJECT_ROOT / 'scripts' / 'support_sanity.py'),
    '--config',
    str(CONFIG),
    '--cases',
    'sub',
    'ins',
    'del',
    'copy',
    '--output',
    str(OUTPUT),
]
print(' '.join(command))
completed = subprocess.run(command, cwd=PROJECT_ROOT, check=True, text=True, capture_output=True, env=env)
print(completed.stdout)
if completed.stderr:
    print(completed.stderr)
```
